# Løsningsforslag med Python

Denne notebooken samler Python-koden fra løsningsforslaget til utvalgte oppgaver (kap. 6, 9, 10 og 11), med hovedvekt på:

- **Binomialkoeffisienten** og binomisk sannsynlighetsfordeling (`math.comb`, `scipy.stats.binom`)

```python
import math
from scipy.stats import binom, norm
```

## Oppgave 6.5c–d

En DJ har en spilleliste med 10 favorittlåter.

- **c)** Antall mulige spillelister når rekkefølgen av alle 10 låtene betyr noe → permutasjoner.
- **d)** Antall måter å velge 3 sanger av 10 på, uten hensyn til rekkefølge → kombinasjoner (binomialkoeffisienten).


```python
# c) Permutasjoner av alle 10 sangene (rekkefølge betyr noe)
antall_permutasjoner = math.factorial(10)
print(antall_permutasjoner)   # 3628800

# d) Kombinasjoner: velge 3 sanger av 10 (rekkefølge betyr ikke noe)
antall_kombinasjoner = math.comb(10, 3)
print(antall_kombinasjoner)   # 120
```

## Oppgave 9.1c

En mynt spinnes, med $P(\text{kron})=0.59$. Vi spinner mynten 100 ganger.
Antall kron $Y$ er binomisk fordelt: $Y\sim\text{Bin}(n=100,\ p=0.59)$.

Vi finner $P(Y=59)$ på to måter: manuelt med binomialkoeffisienten, og med `scipy.stats.binom`.


```python
n, k, p = 100, 59, 0.59

# Manuelt med binomialkoeffisienten
binom_koeff = math.comb(n, k)
P = binom_koeff * p**k * (1-p)**(n-k)
print(binom_koeff)   # 20116440213369968050635175200
print(P)             # 0.0809017550696981

# Samme resultat med scipy sin ferdige binomisk-funksjon
P2 = binom.pmf(k, n, p)
print(P2)            # 0.0809017550696981
```

## Oppgave 10.3a–c

20 flervalgsoppgaver med 4 alternativer hver. Kim Tore gjetter helt tilfeldig, så antall rette
$X\sim\text{Bin}(n=20,\ p=0.25)$.

- **a)** Forventet antall rette
- **b)** $P(X=3)$
- **c)** $P(X>3)$


```python
n, p = 20, 0.25

# a) Forventning
E_X = n * p
print(E_X)                       # 5.0

# b) P(X = 3), her via binomialkoeffisienten "for hånd"
Pb = math.comb(n, 3) * p**3 * (1-p)**(n-3)
print(Pb)                        # 0.13389561519943532
print(binom.pmf(3, n, p))        # samme svar med scipy

# c) P(X > 3) = 1 - P(X <= 3)
Pc = 1 - binom.cdf(3, n, p)
print(Pc)                        # 0.7748439523356865
```

## Oppgave 11.4a–c

Intervjutid $X\sim N(\mu=16,\ \sigma=4)$ minutter.

- **a)** $P(X<10)$
- **b)** Finn $x$ slik at $P(X>x)=0.05$ (Egils intervjutid, 95-persentilen)
- **c)** Gjennomsnitt av $n=4$ intervjuer: $P(\bar{X}<15)$


```python
mu, sigma = 16, 4

# a) P(X < 10)
Pa = norm.cdf(10, loc=mu, scale=sigma)
print(Pa)                         # 0.06680720126885807

# b) Finn x slik at P(X > x) = 0.05, dvs. x er 95-persentilen
x_egil = norm.ppf(0.95, loc=mu, scale=sigma)
print(x_egil)                     # 22.57941450780589

# c) Gjennomsnitt av n=4 intervjuer: SD(Xbar) = sigma/sqrt(n)
n = 4
se = sigma / n**0.5
Pc = norm.cdf(15, loc=mu, scale=se)
print(se, Pc)                     # 2.0   0.3085375387259869
```

## Oppgave 11.6a–e

Fødselsvekt $\sim N(\mu=3365,\ \sigma=577)$ gram.

- **a)** $P(X<4000)$
- **b)** $P(3000<X<4000)$
- **c)** Finn $x$ slik at $P(X>x)=0.20$
- **d)** Gjennomsnitt av $n=5$ nyfødte: $P(\bar{X}>4000)$
- **e)** $x$-verdi som gir $z$-score $=2$


```python
mu, sigma = 3365, 577

# a) P(X < 4000)
print(norm.cdf(4000, mu, sigma))              # 0.8644471746940054

# b) P(3000 < X < 4000)
p_b = norm.cdf(4000, mu, sigma) - norm.cdf(3000, mu, sigma)
print(p_b)                                    # 0.6009439576566514

# c) Finn x slik at P(X > x) = 0.20
print(norm.ppf(1-0.20, mu, sigma))            # 3850.6154517715713

# d) Gjennomsnitt av n=5 nyfødte
n = 5
se = sigma / n**0.5
print(1 - norm.cdf(4000, mu, se))             # 0.006930658999786177

# e) x-verdi som gir z-score = 2
print(mu + 2*sigma)                           # 4519
```

## Oppgave 11.7a–c

Høyde til mannlige OL-deltakere i langrenn $\sim N(\mu=180,\ \sigma=6)$ cm.

- **a)** $P(X<174)$
- **b)** $P(X>191)$
- **c)** $P(174<X<191)$


```python
mu, sigma = 180, 6

# a) P(X < 174)
print(norm.cdf(174, mu, sigma))               # 0.15865525393145707

# b) P(X > 191)
print(1 - norm.cdf(191, mu, sigma))           # 0.03337650758481725

# c) P(174 < X < 191)
p_c = norm.cdf(191, mu, sigma) - norm.cdf(174, mu, sigma)
print(p_c)                                    # 0.8079682384837257
```

## Oppgave 11.8a–e

BMI hos voksne menn i Oslo $\sim N(\mu=26.5,\ \sigma=3.4)$ kg/m$^2$.

- **a)** $P(\text{BMI}>25)$ (overvekt)
- **b)** $P(\text{BMI}<18.5)$ (undervekt)
- **c)** $P(\text{normalvektig})$
- **d)** Gjennomsnitt av $n=5$ menn: $P(\bar{X}>25)$
- **e)** Finn $x$ slik at $P(\text{BMI}>x)=0.12$


```python
mu, sigma = 26.5, 3.4

# a) P(BMI > 25)
p_over25 = 1 - norm.cdf(25, mu, sigma)
print(p_over25)                        # 0.6704573770288774

# b) P(BMI < 18.5)
p_under18_5 = norm.cdf(18.5, mu, sigma)
print(p_under18_5)                     # 0.00931279014370143

# c) P(normalvektig) = 1 - P(over25) - P(under18.5)
print(1 - p_over25 - p_under18_5)      # 0.3202298328274211

# d) Gjennomsnitt av n=5 menn: P(Xbar > 25)
n = 5
se = sigma / n**0.5
print(1 - norm.cdf(25, mu, se))        # 0.8380562341995479

# e) Finn x slik at P(BMI > x) = 0.12
print(norm.ppf(1-0.12, mu, sigma))     # 30.494955093024707
```

## Oppgave 11.8f–g

- **f)** $X=$ antall av 500 menn med BMI$>25$, $X\sim\text{Bin}(500,p)$, tilnærmet normalfordelt siden $np$ og $n(1-p)$ begge er store.
- **g)** $P(X>300)$ med kontinuitetskorreksjon.


```python
mu, sigma = 26.5, 3.4
p = 1 - norm.cdf(25, mu, sigma)   # 0.6704573770288774

# f) X ~ Bin(500, p) tilnærmet normalfordelt
n_menn = 500
E_X = n_menn * p
SD_X = (n_menn * p * (1-p))**0.5
print(E_X, SD_X)                  # 335.2286885144387  10.51057283444711

# g) P(X > 300) med kontinuitetskorreksjon
z = (300.5 - E_X) / SD_X
Pg = 1 - norm.cdf(300.5, E_X, SD_X)
print(Pg)                         # 0.9995237047525459
```
